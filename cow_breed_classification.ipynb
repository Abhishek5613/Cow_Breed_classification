```python
from google.colab import drive
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import pathlib, random
```

```python
# Step 2: Set the dataset path
dataset_path = '/content/drive/MyDrive/COW DATASET 2'
no_bread_classes = os.listdir(dataset_path)
data = pathlib.Path(dataset_path)
CowBread = np.array(sorted([item.name for item in data.glob("*")]))
print(CowBread)
```

```python
def view_random_image(target_dir, target_class):
    target_folder = os.path.join(target_dir, target_class)
    random_image = random.sample(os.listdir(target_folder), 1)
    img = plt.imread(os.path.join(target_folder, random_image[0]))
    plt.imshow(img)
    plt.title(target_class)
    plt.axis("off")
    return img
```

```python
fig, axes = plt.subplots(nrows=4, ncols=4, figsize=(10, 10), subplot_kw={'xticks': [], 'yticks': []})

random_index = np.random.randint(0, len(CowBread), 16)

for i, ax in enumerate(axes.flat):
    ax.imshow(view_random_image(dataset_path, CowBread[random_index[i]]))
    ax.set_title(CowBread[random_index[i]])

plt.show()
```
